# Data Cleaning

In this stage the dataset is cleaned and ready for exploratory data analysis (EDA) and further preprocessing. The following steps are taken:

- Removing customerID: `customerID` is a row identifier and has no meaningful information for analysis and modeling.
- Converting `TotalCharges` from string to float: `Total Charges` is a numeric feature but stored as a string.
- Finding and handling missing values: Even though the dataset has no standard missing values it may have hidden missing values in `Total Charges`.

```python
import pandas as pd
import numpy as np
```

```python
file_name = '../data/raw/WA_Fn-UseC_-Telco-Customer-Churn.csv'
df = pd.read_csv(file_name)
```

### Making a Copy of Dataset

A working copy of the raw dataset is taken to retain the raw dataset. All the cleaning process will be done on the copy of dataset.

```python
df_clean = df.copy()
```

### Remove Customer ID

`customerID` is a unique identifier assigned to each customer. Since it does not contain predictive information about customer churn, it will be removed before training.

```python
df_clean = df_clean.drop('customerID', axis = 1)
```

```python
df_clean.columns
```

### Convert Total Charges to numeric

The `TotalCharges` feature represents the total amount paid by each customer. Since this feature contains numerical information but its stored as a string, it needs to be converted into a numeric format before model training.

```python
df_clean[df_clean['TotalCharges'] == " "]
```

```python
df_clean['TotalCharges'] = df_clean['TotalCharges'].replace(" ", np.nan)
```

```python
df_clean['TotalCharges'] = pd.to_numeric((df_clean['TotalCharges']))
```

```python
df_clean['TotalCharges'].dtype
```

### Handling Missing Values in TotalCharges

The missing values in `TotalCharges` belong to customers with zero tenure. Since these customers have not completed any months of subscription, their total charges are considered zero. Therefore, missing values in `TotalCharges` are replaced with 0.

```python
df_clean['TotalCharges'] = df_clean['TotalCharges'].fillna(0)
```

```python
df_clean.isnull().sum()
```

```python
df_clean.info()
```

### Save the Cleaned Dataset

The cleaned dataset will be saved into a csv file and used for EDA and modeling.

```python
df_clean.to_csv('../data/processed/TelcoCustomerChurnCleaned.csv', index = False)
```

## Summary of Data Cleaning

The dataset is cleaned and is now ready for analysis and modeling.

- `customerID` is removed.
- `Total Charges` is converted to a numeric features.
- The missing values in `Total Charges` are detected and replaced by `0`.
