## Initial Data Quality Observations

During the initial exploration of the dataset, some potential data quality issues were idetified:

- `customerID` is an identifier column and does not provide meaningful predictive information. It may be removed during feature engineering.
- `TotalCharges` represents a numerical feature, but it is stored as a string data type and requires conversion before modeling.

```python
df.isnull().sum()
```

## Missing Values Check

No missing values were detected using standard missing value detection methods.

However, the `TotalCharges` feature requires further investigation because it is stored as a string data type and may contain non-numeric values.

```python
df['Churn'].value_counts(normalize=True)
```

## Target Variable Distribution

The target variable `Churn` is moderately imbalanced.

Approximately 73.4% of customers have not churned, while 26.5% of customers have churned.

Because the dataset is imbalanced, accuracy alone may not be a reliable evaluation metric. Therefore additional metrics such as precision, recall and F1-score will be considered.
